```python
Example Quadratic equation solver:<br>
Given coefficients $a,b,c$ of equations $ax^2+bx+c=0$, then to solve we check for the discriminant, $D=b^2-4ac$<br>
So if $D==0$, then roots are equal, else if $D>0$, then roots are distinct and real, else ($D<0$), then roots are complex
```

```python
def main():
  a=int(input("a = "))
  b=int(input("b = "))
  c=int(input("c = "))
  D=b**2-4*a*c
  if D==0:
    print("Roots are equal")
  elif D>0:
    print("Roots are distinct and real")
  else:
    print("Roots are imaginary")
```

```python
main()
```

Finding maximum of 3 numbers

```python
def main():
  x1=int(input("Enter 1st number: "))
  x2=int(input("Enter 2nd number: "))
  x3=int(input("Enter 3rd number: "))
  if x1>x2>x3:
    print(f"{x1} is largest")
  elif x1<x2>x3:
    print(f"{x2} is largest")
  else:
    print(f"{x3} is largest")
```

**Nesting of if-else**

```python
def main():
  x1=int(input("Enter 1st number: "))
  x2=int(input("Enter 2nd number: "))
  x3=int(input("Enter 3rd number: "))
  if x1>x2:
    if x1>x3:
      print(f"{x1} is largest")
    else:
      print(f"{x3} is largest")
  else:
    if x2>x3:
      print(f"{x2} is largest")
    else:
      print(f"{x3} is largest")
```

```python
main()
```

Take 3 side of a trgainle,print whether it is scalene ,isoceles or equilateral

```python
def main():
  a = float(input("Enter side a:"))
  b = float(input("Enter side b:"))
  c = float(input("Enter side c:"))
  if a==b==c:
    print("Equilateral")
  elif a!=b!=c:
    print(f"{a},{b},{c} are sides of scalene triangle")
  else:
    print("Isoceles triangle")
```

```python
main()
```
